```python
import pandas as pd
import numpy as np

# 1. Load Dataset
file_path = r"D:/dax/PAIP Dataset.csv"
df = pd.read_csv(file_path)
```

```python
print("\n--- DATA TYPES & NON-NULL COUNTS ---")
print(df.info())
```

```python
# 3. Missing Data Analysis
print("\n--- MISSING VALUES SUMMARY ---")
missing_val = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing_Count': missing_val,
    'Missing_Percentage (%)': missing_pct.round(2)
})

# Filter to show columns with missing values (or all if none are missing)
cols_with_missing = missing_df[missing_df['Missing_Count'] > 0]

if len(cols_with_missing) > 0:
    print(cols_with_missing.sort_values(by='Missing_Count', ascending=False))
else:
    print("Good news! No standard NaN/null values found across any columns.")
```

```python
# 4. Check for Duplicate Rows
duplicate_count = df.duplicated().sum()
print(f"\n--- DUPLICATE ROWS ---")
print(f"Total duplicate rows: {duplicate_count}")
```

```python
# 5. Check for Non-Standard Missing Values (Blank spaces, 'N/A', '-', etc.)
print("\n--- POTENTIAL NON-STANDARD MISSING VALUES ---")
non_standard_found = False
for col in df.select_dtypes(include=['object']).columns:
    # Check for empty strings, whitespace, or placeholder symbols
    invalid_mask = df[col].astype(str).str.strip().isin(['', ' ', '-', 'N/A', 'nan', 'null', 'None', '?'])
    count = invalid_mask.sum()
    if count > 0:
        print(f"Column [{col}] has {count} entries with potential non-standard missing representations.")
        non_standard_found = True

if not non_standard_found:
    print("No non-standard placeholders ('-', 'N/A', empty strings) detected in text columns.")
```

```python
# 6. Check for Special Formatting Issues in Numeric Columns
print("\n--- STRING-FORMATTED NUMERIC COLUMNS CHECK ---")
numeric_cols = [
    'Plant_Capacity_m3_day', 'Population_Served', 'Customer_Accounts', 
    'Production_m3', 'Domestic_Billed_m3', 'Commercial_Billed_m3', 
    'Industrial_Billed_m3', 'Energy_kWh', 'Chemical_Cost_RM', 
    'Maintenance_Cost_RM', 'Total_Billed_m3', 'NRW_m3', 'Billed_Revenue_RM', 
    'Energy_Cost_RM', 'Total_Operating_Cost_RM'
]

for col in numeric_cols:
    if col in df.columns and df[col].dtype == 'object':
        # Check if commas exist in data preventing direct numeric conversion
        comma_count = df[col].astype(str).str.contains(',').sum()
        print(f"Column [{col}] is stored as text/object (Commas present in {comma_count} rows). Needs type casting.")
```

```python
import pandas as pd
import numpy as np

# 1. Load Dataset
input_path = r"D:\dax\PAIP Dataset.csv"
output_path = r"D:\dax\PAIP_Dataset_Cleaned.csv"

df = pd.read_csv(input_path)
print(f"Original Dataset Shape: {df.shape}")

# 2. Parse Date Column
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# 3. Clean String-Formatted Numeric Columns (Remove Commas & Symbols)
comma_cols = [
    'Plant_Capacity_m3_day', 'Population_Served', 'Customer_Accounts', 
    'Production_m3', 'Domestic_Billed_m3', 'Commercial_Billed_m3', 
    'Industrial_Billed_m3', 'Energy_kWh', 'Chemical_Cost_RM', 
    'Maintenance_Cost_RM', 'Total_Billed_m3', 'NRW_m3', 'Billed_Revenue_RM', 
    'Energy_Cost_RM', 'Total_Operating_Cost_RM', 'Physical_Loss_m3', 
    'Commercial_Loss_m3', 'NRW_per_km_m3'
]

for col in comma_cols:
    if col in df.columns and df[col].dtype == 'object':
        df[col] = df[col].astype(str).str.replace(',', '').str.strip()
        df[col] = pd.to_numeric(df[col], errors='coerce')

# 4. Clean Percentage Columns (Remove '%' signs if present)
pct_cols = ['Capacity_Utilization_pct', 'NRW_Percentage', 'Operating_Margin_pct', 'Physical_Loss_Proportion']

for col in pct_cols:
    if col in df.columns and df[col].dtype == 'object':
        df[col] = df[col].astype(str).str.replace('%', '').str.strip()
        df[col] = pd.to_numeric(df[col], errors='coerce')

# 5. Handle Missing Values via Plant/District Group Median Imputation
# Impute Raw_Water_Turbidity_NTU (group by District, fallback to overall median)
df['Raw_Water_Turbidity_NTU'] = df.groupby('District')['Raw_Water_Turbidity_NTU'].transform(
    lambda x: x.fillna(x.median())
)
df['Raw_Water_Turbidity_NTU'] = df['Raw_Water_Turbidity_NTU'].fillna(df['Raw_Water_Turbidity_NTU'].median())

# Impute Average_Pressure_bar (group by Plant_Name, fallback to overall median)
df['Average_Pressure_bar'] = df.groupby('Plant_Name')['Average_Pressure_bar'].transform(
    lambda x: x.fillna(x.median())
)
df['Average_Pressure_bar'] = df['Average_Pressure_bar'].fillna(df['Average_Pressure_bar'].median())

# 6. Verify Post-Cleaning Integrity
print("\n--- POST-CLEANING INTEGRITY CHECK ---")
print(f"Remaining Missing Values:\n{df[['Raw_Water_Turbidity_NTU', 'Average_Pressure_bar']].isnull().sum()}")
print(f"Total Non-Numeric Columns Left: {len(df.select_dtypes(include=['object']).columns)}")
print(f"Non-Numeric Columns List: {list(df.select_dtypes(include=['object']).columns)}")

# 7. Save Cleaned Dataset
df.to_csv(output_path, index=False)
print(f"\nCleaned dataset successfully saved to: {output_path}")
```
